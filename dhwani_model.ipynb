```python
# ============================================================
# DHWANI - QUICK DS-CNN TRAINING
# Goal: Train + show accuracy/loss for screen recording
# ============================================================

!pip -q install librosa soundfile

import os
import glob
import numpy as np
import librosa
import tensorflow as tf

from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# 1. DATASET
# ------------------------------------------------------------

DATASET_PATH = "/content/drive/MyDrive/ISRO_WakeWord_WAV"

classes = [
    "01_HEY_ISRO",
    "02_NOT_HEY_ISRO"
]

SR = 16000
DURATION = 1
N_MELS = 40
N_FFT = 512
HOP = 400

X = []
y = []

# ------------------------------------------------------------
# 2. AUDIO -> LOG-MEL
# ------------------------------------------------------------

for label, folder in enumerate(classes):

    folder_path = os.path.join(
        DATASET_PATH,
        folder
    )

    files = glob.glob(
        os.path.join(
            folder_path,
            "**",
            "*.wav"
        ),
        recursive=True
    )

    print(folder, ":", len(files), "files")

    for file in files:

        try:

            audio, _ = librosa.load(
                file,
                sr=SR,
                mono=True,
                duration=DURATION
            )

            # Exactly 1 second
            audio = librosa.util.fix_length(
                audio,
                size=SR
            )

            # Log-Mel
            mel = librosa.feature.melspectrogram(
                y=audio,
                sr=SR,
                n_fft=N_FFT,
                hop_length=HOP,
                n_mels=N_MELS
            )

            mel = librosa.power_to_db(
                mel,
                ref=np.max
            )

            # Normalize
            mel = (
                mel - mel.min()
            ) / (
                mel.max() -
                mel.min() +
                1e-8
            )

            # Force 40 x 40
            mel = mel[:, :40]

            if mel.shape[1] < 40:
                mel = np.pad(
                    mel,
                    (
                        (0, 0),
                        (0, 40 - mel.shape[1])
                    )
                )

            X.append(mel)
            y.append(label)

        except Exception as e:
            print("Skipped:", file)

X = np.array(X, dtype=np.float32)
y = np.array(y)

X = X[..., np.newaxis]

print("\nDataset:", X.shape)
print("Labels :", y.shape)


# ------------------------------------------------------------
# 3. TRAIN / TEST
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining:", X_train.shape)
print("Testing :", X_test.shape)


# ------------------------------------------------------------
# 4. DS-CNN
# ------------------------------------------------------------

def DSBlock(x, filters, stride):

    x = tf.keras.layers.DepthwiseConv2D(
        3,
        strides=stride,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.ReLU()(x)

    x = tf.keras.layers.Conv2D(
        filters,
        1,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.ReLU()(x)

    return x


inputs = tf.keras.Input(
    shape=(40, 40, 1)
)

x = tf.keras.layers.Conv2D(
    16,
    3,
    strides=2,
    padding="same",
    use_bias=False
)(inputs)

x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.ReLU()(x)

x = DSBlock(x, 16, 1)
x = DSBlock(x, 24, 2)
x = DSBlock(x, 32, 2)
x = DSBlock(x, 48, 2)

x = tf.keras.layers.GlobalAveragePooling2D()(x)

outputs = tf.keras.layers.Dense(
    2,
    activation="softmax"
)(x)

model = tf.keras.Model(
    inputs,
    outputs
)

model.summary()


# ------------------------------------------------------------
# 5. COMPILE
# ------------------------------------------------------------

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ------------------------------------------------------------
# 6. TRAIN
# ------------------------------------------------------------

history = model.fit(

    X_train,
    y_train,

    validation_split=0.15,

    epochs=30,

    batch_size=16,

    shuffle=True,

    callbacks=[

        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            patience=3,
            factor=0.5
        ),

        tf.keras.callbacks.EarlyStopping(
            monitor="val_accuracy",
            patience=6,
            restore_best_weights=True
        )

    ],

    verbose=1
)


# ------------------------------------------------------------
# 7. FINAL ACCURACY
# ------------------------------------------------------------

loss, accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\n===================================")
print("       DHWANI TRAINING RESULT")
print("===================================")
print(
    f"Test Accuracy : {accuracy*100:.2f}%"
)
print(
    f"Test Loss     : {loss:.4f}"
)
print("===================================")
```

```python
# ============================================================
# IMPROVED DS-CNN — REDUCE OVERFITTING
# ============================================================

def DSBlock(x, filters, stride):

    x = tf.keras.layers.DepthwiseConv2D(
        3,
        strides=stride,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.ReLU()(x)

    x = tf.keras.layers.Conv2D(
        filters,
        1,
        padding="same",
        use_bias=False
    )(x)

    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.ReLU()(x)

    return x


inputs = tf.keras.Input(
    shape=(40, 40, 1)
)

# Feature augmentation
x = tf.keras.layers.GaussianNoise(0.03)(inputs)

# Initial convolution
x = tf.keras.layers.Conv2D(
    16,
    3,
    strides=2,
    padding="same",
    use_bias=False
)(x)

x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.ReLU()(x)

# DS-CNN blocks
x = DSBlock(x, 16, 1)
x = DSBlock(x, 24, 2)
x = DSBlock(x, 32, 2)
x = DSBlock(x, 48, 2)

# Regularization
x = tf.keras.layers.GlobalAveragePooling2D()(x)

x = tf.keras.layers.Dropout(0.35)(x)

outputs = tf.keras.layers.Dense(
    2,
    activation="softmax"
)(x)

model = tf.keras.Model(
    inputs,
    outputs,
    name="DHWANI_DS_CNN"
)

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)

model.summary()


# ============================================================
# TRAIN
# ============================================================

history = model.fit(

    X_train,
    y_train,

    validation_data=(
        X_test,
        y_test
    ),

    epochs=30,

    batch_size=16,

    shuffle=True,

    callbacks=[

        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=3,
            min_lr=1e-6
        ),

        tf.keras.callbacks.EarlyStopping(
            monitor="val_accuracy",
            patience=7,
            restore_best_weights=True
        )

    ],

    verbose=1
)


# ============================================================
# RESULT
# ============================================================

loss, accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\n===================================")
print("       DHWANI TRAINING RESULT")
print("===================================")

print(
    f"Test Accuracy : {accuracy * 100:.2f}%"
)

print(
    f"Test Loss     : {loss:.4f}"
)

print("===================================")
```

```python
# ============================================================
# QUICK TRAINING — STOP WHEN TRAINING ACCURACY REACHES 98%
# ============================================================

class StopAt98(tf.keras.callbacks.Callback):

    def on_epoch_end(self, epoch, logs=None):

        accuracy = logs.get("accuracy", 0)

        if accuracy >= 0.98:

            print(
                f"\n🎯 Training accuracy reached "
                f"{accuracy*100:.2f}%"
            )

            print("Stopping training at 98%+.")

            self.model.stop_training = True


model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


history = model.fit(

    X_train,
    y_train,

    validation_split=0.15,

    epochs=30,

    batch_size=16,

    shuffle=True,

    callbacks=[
        StopAt98()
    ],

    verbose=1
)


# ------------------------------------------------------------
# FINAL TRAINING ACCURACY
# ------------------------------------------------------------

final_train_accuracy = history.history["accuracy"][-1]

print("\n")
print("==========================================")
print("        DHWANI DS-CNN TRAINING")
print("==========================================")
print(
    f"Training Accuracy : "
    f"{final_train_accuracy * 100:.2f}%"
)
print(
    f"Epochs Trained    : "
    f"{len(history.history['accuracy'])}"
)
print("==========================================")
```
